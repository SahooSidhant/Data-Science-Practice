## HYPER PARAMETER TUNING AND CROSS VALIDATION

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, GridSearchCV

# Create model
model = LogisticRegression(max_iter=1000)

# Valid parameter combinations
params = [
    {
        'solver': ['lbfgs', 'newton-cg', 'newton-cholesky', 'sag'],
        'penalty': ['l2'],
        'C': [100, 10, 1.0, 0.1, 0.01]
    },
    {
        'solver': ['liblinear'],
        'penalty': ['l1', 'l2'],
        'C': [100, 10, 1.0, 0.1, 0.01]
    },
    {
        'solver': ['saga'],
        'penalty': ['l1', 'l2', 'elasticnet'],
        'C': [100, 10, 1.0, 0.1, 0.01],
        'l1_ratio': [0.5]   # Required only for elasticnet
    }
]

# Cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Grid Search
grid = GridSearchCV(
    estimator=model,
    param_grid=params,
    scoring='accuracy',
    cv=cv,
    n_jobs=-1
)

# Train
grid.fit(X_train, y_train)

# Best parameters
print("Best Score:", grid.best_score_)
print("Best Parameters:", grid.best_params_)
```

## GridSearch CV

```python
y_pred=grid.predict(X_test)
```

```python
score = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
print(score)
print(cm)
print(classification_report(y_test,y_pred))
```

```python
params = {
    'penalty': penalty,
    'C': c_values,
    'class_weight': class_weight,
    'solver': solver
}
```

```python
from sklearn.model_selection import StratifiedKFold, GridSearchCV
cv = StratifiedKFold()
grid = GridSearchCV(estimator=model, param_grid=params, scoring='accuracy', cv=cv)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
y_pred=grid.predict(X_test)
```

```python
score = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
print(score)
print(cm)
print(classification_report(y_test,y_pred))
```

# Classification Metrics Formulas

## 1. False Positive Rate (FPR)

$$
\text{FPR} = \frac{\text{FP}}{\text{N}} = \frac{\text{FP}}{\text{FP} + \text{TN}}
$$

### Variables Defined:
* **$\text{FP}$**: Number of false positives
* **$\text{TN}$**: Number of true negatives
* **$\text{N}$**: Total number of actual negatives ($\text{FP} + \text{TN}$)

---

## 2. True Positive Rate (TPR)
*Also known as **sensitivity**, **recall**, or **hit rate**.*

$$
\text{TPR} = \frac{\text{TP}}{\text{P}} = \frac{\text{TP}}{\text{TP} + \text{FN}} = 1 - \text{FNR}
$$

### Variables Defined:
* **$\text{TP}$**: Number of true positives
* **$\text{FN}$**: Number of false negatives
* **$\text{P}$**: Total number of actual positives ($\text{TP} + \text{FN}$)
* **$\text{FNR}$**: False Negative Rate

```python
## calculate ROC Curve
dummy_fpr,dummy_tpr,_ = roc_curve(y_test,dummy_model_prob)
model_fpr,model_tpr,thresholds = roc_curve(y_test,model_prob)
```

```python
thresholds
```

```python
model_fpr, model_tpr
```

```python
## plot the roc curve for the model
pyplot.plot(dummy_fpr,dummy_tpr,linestyle='--',label='Dummy Model')
pyplot.plot(model_fpr,model_tpr,marker='.',label='Logistic')
# axis label
pyplot.xlabel('False Positive Rate')
pyplot.ylabel('True Positive Rate')
# show the legend
pyplot.legend()
# show the output
pyplot.show()
```

```python
# plot the roc curve for the model
import numpy as np
fig = pyplot.figure(figsize=(20,50))
ax = fig.add_subplot(111)
pyplot.plot(dummy_fpr, dummy_tpr, linestyle='--', label='Dummy Model')
pyplot.plot(model_fpr, model_tpr, marker='.', label='Logistic')
for xyz in zip(model_fpr, model_tpr, thresholds):
    ax.annotate('%s' % np.round(xyz[2],2), xy=(xyz[0],xyz[1]))
# axis labels
pyplot.xlabel('False Positive Rate')
pyplot.ylabel('True Positive Rate')
# show the legend
pyplot.legend()
# show the plot
pyplot.show()
```
